```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
```

```python
dataset = pd.read_csv(r"C:\Users\cp lab 25\Downloads\spam1.csv")
```

```python
# Display Dataset Information
print(dataset.head())
print(dataset.info())

# Keep only the first two columns (v1 and v2)
dataset = dataset.iloc[:, :2]

# Rename columns
dataset.columns = ['Category', 'Message']

# Remove missing values
dataset = dataset.dropna()

# Plot Spam and Ham count
plt.figure(figsize=(6,4))
sns.countplot(x='Category', data=dataset)
plt.title("Spam and Ham Count")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python

# Split data into features and labels
X = dataset['Message'].values
y = dataset['Category'].values

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=0
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X_train = cv.fit_transform(X_train)
X_test = cv.transform(X_test)

from sklearn.svm import SVC
classifier = SVC(kernel='rbf', random_state=0)
classifier.fit(X_train, y_train)
```

```python
print(classifier.score(X_test, y_test)) 
```
